```python
class Human:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def check_adult(self):
    if self.age >= 20:
      print(f"{self.name}は大人です。")
    else:
      print(f"{self.name}は大人ではありません。")

human_1 = Human("侍一郎", 21)
human_2 = Human("侍ニ郎", 20)
human_3 = Human("侍三郎", 19)

array = [human_1, human_2, human_3]

for human in array:
  human.check_adult()


```
